## 1. Variables and data types

```python
country = "Sri Lanka"
inflation = 12.5
gdp_growth = -2.3
exchange_rate = 300

print(country, inflation, gdp_growth, exchange_rate)
```

## 2. Lists and economic data

```python
gdp_growth_rates = [3.5, 5.2, -7.8, 2.1]

average_growth = sum(gdp_growth_rates) / len(gdp_growth_rates)
maximum_growth = max(gdp_growth_rates)
minimum_growth = min(gdp_growth_rates)

print(average_growth, maximum_growth, minimum_growth)
```

## 3. Conditional logic

```python
inflation = 12

if inflation < 5:
    print("Low inflation")
elif inflation <= 15:
    print("Moderate inflation")
else:
    print("High inflation")
```

## 4. Loops and automation

```python
for growth in gdp_growth_rates:
    if growth >= 0:
        print(growth, "Positive growth")
    else:
        print(growth, "Recession year")
```

## 5. Functions and reusable code

```python
def percentage_change(old_value, new_value):
    return ((new_value - old_value) / old_value) * 100
```

```python
gdp_change = percentage_change(5000, 5400)
cpi_change = percentage_change(200, 240)
exchange_rate_change = percentage_change(300, 330)

print("GDP change:", gdp_change)
print("CPI change:", cpi_change)
print("Exchange-rate change:", exchange_rate_change)
```
