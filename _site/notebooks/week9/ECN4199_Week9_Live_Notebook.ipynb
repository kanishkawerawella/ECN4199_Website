## 3. Classification Variables

`High_Risk` is the target variable. Income, debt, savings, late payments, employment, and debt-to-income ratio are features.


```python
data.describe()
data["High_Risk"].value_counts()
```

## 4. Visualize Risk Patterns


```python
plt.scatter(data["Debt_to_Income"], data["Savings"], c=data["High_Risk"])
plt.xlabel("Debt-to-income ratio")
plt.ylabel("Savings")
plt.title("Credit Risk Groups")
plt.show()
```

## 5. Train a Logistic Regression Classifier


```python
features = ["Income", "Debt", "Savings", "Late_Payments", "Employed", "Debt_to_Income"]
X = data[features]
y = data["High_Risk"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
predictions = model.predict(X_test)
print("Accuracy:", round(accuracy_score(y_test, predictions), 3))
print(confusion_matrix(y_test, predictions))
print(classification_report(y_test, predictions))
```

## 6. Interpret Errors

False positives and false negatives have different economic consequences. A lender, bank, or policymaker must explain which error is more costly.


```python
results = X_test.copy()
results["Actual_High_Risk"] = y_test.values
results["Predicted_High_Risk"] = predictions
results
```
