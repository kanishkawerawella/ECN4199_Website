## Descriptive Statistics and Data Visualization for Economics

Save a copy of this notebook before editing: **File -> Save a copy in Drive**.


```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/kanishkawerawella/ECN4199_Website/main/data/sri_lanka_macro.csv"
data = pd.read_csv(url)
data["Exchange_Rate_Change"] = data["Exchange_Rate"].pct_change() * 100

data
```

## 1. Descriptive Statistics

```python
data.describe()
```

```python
mean_inflation = data["Inflation"].mean()
median_inflation = data["Inflation"].median()
std_inflation = data["Inflation"].std()

mean_inflation, median_inflation, std_inflation
```

## 2. Histogram

```python
plt.hist(data["Inflation"], bins=6)
plt.title("Distribution of Inflation")
plt.xlabel("Inflation")
plt.ylabel("Frequency")
plt.show()
```

## 3. Line Graph

```python
plt.plot(data["Year"], data["Inflation"], marker="o")
plt.title("Inflation Trends")
plt.xlabel("Year")
plt.ylabel("Inflation")
plt.show()
```

## 4. Correlation

```python
data["Inflation"].corr(data["Exchange_Rate_Change"])
```
