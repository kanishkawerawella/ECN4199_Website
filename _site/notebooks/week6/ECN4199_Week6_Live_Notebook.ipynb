This notebook turns Week 5 evidence into an interactive dashboard.

## 1. Dashboard Question

Write the main question your dashboard should answer. Example: How did Sri Lanka's macroeconomic conditions change across the period?

```python
import pandas as pd
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go

url = "https://raw.githubusercontent.com/kanishkawerawella/ECN4199_Website/main/data/sri_lanka_macro.csv"
data = pd.read_csv(url)

data["Exchange_Rate_Change"] = data["Exchange_Rate"].pct_change() * 100
data["Recession"] = data["GDP_Growth"] < 0

data
```

## 2. Interactive Inflation Chart

```python
fig = px.line(
    data,
    x="Year",
    y="Inflation",
    markers=True,
    title="Inflation Over Time",
    hover_data=["GDP_Growth", "Exchange_Rate"]
)

fig.show()
```

## 3. Interactive GDP Growth Chart

```python
fig = px.bar(
    data,
    x="Year",
    y="GDP_Growth",
    color="GDP_Growth",
    title="GDP Growth by Year"
)

fig.show()
```

## 4. Interactive Relationship Chart

```python
fig = px.scatter(
    data,
    x="Inflation",
    y="Exchange_Rate_Change",
    hover_name="Year",
    size="Unemployment",
    title="Inflation and Exchange Rate Change"
)

fig.show()
```

## 5. Multi-Panel Monitoring Dashboard

```python
fig = make_subplots(
    rows=3,
    cols=1,
    subplot_titles=("Inflation", "GDP Growth", "Exchange Rate")
)

fig.add_trace(
    go.Scatter(x=data["Year"], y=data["Inflation"], mode="lines+markers", name="Inflation"),
    row=1,
    col=1
)

fig.add_trace(
    go.Bar(x=data["Year"], y=data["GDP_Growth"], name="GDP Growth"),
    row=2,
    col=1
)

fig.add_trace(
    go.Scatter(x=data["Year"], y=data["Exchange_Rate"], mode="lines+markers", name="Exchange Rate"),
    row=3,
    col=1
)

fig.update_layout(
    height=850,
    title_text="Macroeconomic Monitoring Dashboard",
    showlegend=False
)

fig.show()
```
