## Interactive Dashboards and Data Storytelling for Economics

Save a copy of this notebook before editing: **File -> Save a copy in Drive**.


```python
import pandas as pd
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go

url = "https://raw.githubusercontent.com/kanishkawerawella/ECN4199_Website/main/data/sri_lanka_macro.csv"
data = pd.read_csv(url)
data["Exchange_Rate_Change"] = data["Exchange_Rate"].pct_change() * 100

data
```

## 1. Interactive Inflation Line Graph

```python
fig = px.line(data, x="Year", y="Inflation", markers=True, title="Inflation Trends")
fig.show()
```

## 2. Interactive GDP Growth Bar Chart

```python
fig = px.bar(data, x="Year", y="GDP_Growth", color="GDP_Growth", title="GDP Growth by Year")
fig.show()
```

## 3. Interactive Scatter Plot

```python
fig = px.scatter(data, x="Inflation", y="Exchange_Rate_Change", hover_name="Year", title="Inflation and Exchange Rate Change")
fig.show()
```

## 4. Mini Dashboard

```python
fig = make_subplots(rows=3, cols=1, subplot_titles=("Inflation", "GDP Growth", "Exchange Rate"))

fig.add_trace(go.Scatter(x=data["Year"], y=data["Inflation"], mode="lines+markers", name="Inflation"), row=1, col=1)
fig.add_trace(go.Bar(x=data["Year"], y=data["GDP_Growth"], name="GDP Growth"), row=2, col=1)
fig.add_trace(go.Scatter(x=data["Year"], y=data["Exchange_Rate"], mode="lines+markers", name="Exchange Rate"), row=3, col=1)

fig.update_layout(height=850, title_text="Macroeconomic Monitoring Dashboard")
fig.show()
```
