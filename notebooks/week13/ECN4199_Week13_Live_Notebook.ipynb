## 1. Setup


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
plt.style.use("seaborn-v0_8-whitegrid")
```

## 2. Load the Dataset


```python
from io import StringIO

url = "https://raw.githubusercontent.com/kanishkawerawella/ECN4199_Website/main/data/economic_intelligence_project.csv"
try:
    data = pd.read_csv(url)
    print("Using online dataset from GitHub.")
except Exception:
    embedded_csv = """Region,Employment_Rate,Median_Income,Digital_Access,SME_Density,Tourism_Index,Policy_Priority_Score
Western,88.6,60500,52.6,14.1,56,67.9
Central,89.2,69000,57.2,16.2,72,70.8
Southern,92.2,77500,61.8,18.3,88,58.7
Northern,92.8,86000,66.4,20.4,44,61.6
Eastern,86.4,94500,71.0,22.5,60,64.5
North Western,89.4,103000,75.6,24.6,76,52.4
North Central,90.0,111500,80.2,26.7,92,55.3
Uva,90.6,120000,84.8,28.8,48,58.2
Sabaragamuwa,93.6,128500,89.4,30.9,64,46.1"""
    data = pd.read_csv(StringIO(embedded_csv))
    print("Using embedded dataset fallback.")

data.head()
```

## 3. Inspect the Capstone Dataset


```python
data.describe()
data.head()
```

## 4. Build an Economic Profile


```python
data.sort_values("Policy_Priority_Score", ascending=False)[["Region", "Policy_Priority_Score", "Employment_Rate", "Digital_Access"]]
```

## 5. Visualize Regional Differences


```python
plt.scatter(data["Digital_Access"], data["Median_Income"], s=data["SME_Density"]*8)
for _, row in data.iterrows():
    plt.text(row["Digital_Access"], row["Median_Income"], row["Region"], fontsize=8)
plt.xlabel("Digital Access")
plt.ylabel("Median Income")
plt.title("Regional Digital Access and Income")
plt.show()
```

## 6. Segment Regions


```python
features = ["Employment_Rate", "Median_Income", "Digital_Access", "SME_Density", "Tourism_Index", "Policy_Priority_Score"]
scaled = StandardScaler().fit_transform(data[features])
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
data["Segment"] = kmeans.fit_predict(scaled)
data[["Region", "Segment"] + features]
```

## 7. Executive Summary Draft


```python
for segment, group in data.groupby("Segment"):
    print("Segment", segment)
    print("Regions:", ", ".join(group["Region"]))
    print(group[features].mean().round(2))
    print()
```
