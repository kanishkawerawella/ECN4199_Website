## 1. Analytical Questions

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/kanishkawerawella/ECN4199_Website/main/data/sri_lanka_macro.csv"
data = pd.read_csv(url)

data["Exchange_Rate_Change"] = data["Exchange_Rate"].pct_change() * 100
data["Recession"] = data["GDP_Growth"] < 0
data["Low_Reserves"] = data["Reserves"] < 4000

def classify_inflation(rate):
    if rate < 5:
        return "Low"
    elif rate <= 15:
        return "Moderate"
    else:
        return "High"

data["Inflation_Category"] = data["Inflation"].apply(classify_inflation)
data
```

## 2. Descriptive Statistics

```python
analysis_vars = ["Inflation", "GDP_Growth", "Exchange_Rate_Change", "Reserves", "Unemployment"]
data[analysis_vars].describe()
```

```python
print("Mean inflation:", data["Inflation"].mean())
print("Median inflation:", data["Inflation"].median())
print("Maximum inflation:", data["Inflation"].max())
print("Lowest GDP growth:", data["GDP_Growth"].min())
print("Lowest reserves:", data["Reserves"].min())
```

```python
data[analysis_vars].std()
```

## 4. Distribution Analysis

```python
plt.hist(data["Inflation"], bins=6)
plt.title("Distribution of Inflation")
plt.xlabel("Inflation Rate")
plt.ylabel("Number of Years")
plt.show()
```

## 5. Trend Analysis

```python
plt.plot(data["Year"], data["Inflation"], marker="o")
plt.title("Inflation Over Time")
plt.xlabel("Year")
plt.ylabel("Inflation Rate")
plt.show()
```

Write one sentence identifying the main trend or turning point.

## 6. Comparison Analysis

```python
plt.bar(data["Year"], data["GDP_Growth"])
plt.axhline(0, color="black", linewidth=1)
plt.title("GDP Growth by Year")
plt.xlabel("Year")
plt.ylabel("GDP Growth")
plt.show()
```

```python
data["Inflation_Category"].value_counts().plot(kind="bar")
plt.title("Number of Years by Inflation Category")
plt.xlabel("Inflation Category")
plt.ylabel("Number of Years")
plt.show()
```

Write one sentence interpreting expansion, recession, or inflation categories.

## 7. Relationship Analysis

```python
plt.scatter(data["Inflation"], data["Exchange_Rate_Change"])
plt.title("Inflation and Exchange Rate Change")
plt.xlabel("Inflation")
plt.ylabel("Exchange Rate Change")
plt.show()
```

```python
correlation = data["Inflation"].corr(data["Exchange_Rate_Change"])
correlation
```
