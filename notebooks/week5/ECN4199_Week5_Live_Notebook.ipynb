## 1. Analytical Questions

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/kanishkawerawella/ECN4199_Website/main/data/sri_lanka_macro.csv"
data = pd.read_csv(url)

data["Exchange_Rate_Change"] = data["Exchange_Rate"].pct_change() * 100
data["Recession"] = data["GDP_Growth"] < 0

data
```

## 2. Descriptive Statistics

```python
analysis_vars = ["Inflation", "GDP_Growth", "Exchange_Rate_Change", "Reserves", "Unemployment"]
data[analysis_vars].describe()
```

```python
print("Mean inflation:", data["Inflation"].mean())
print("Median inflation:", data["Inflation"].median())
print("Maximum inflation:", data["Inflation"].max())
print("Lowest GDP growth:", data["GDP_Growth"].min())
```

### Interpretation

Which statistic best shows macroeconomic instability? Why is the average not enough?

## 3. Visual Evidence

```python
plt.plot(data["Year"], data["Inflation"], marker="o")
plt.title("Inflation Over Time")
plt.xlabel("Year")
plt.ylabel("Inflation Rate")
plt.show()
```

Write one sentence interpreting the inflation trend.

```python
plt.bar(data["Year"], data["GDP_Growth"])
plt.axhline(0, color="black", linewidth=1)
plt.title("GDP Growth by Year")
plt.xlabel("Year")
plt.ylabel("GDP Growth")
plt.show()
```

Write one sentence interpreting the recession years.

```python
plt.scatter(data["Inflation"], data["Exchange_Rate_Change"])
plt.title("Inflation and Exchange Rate Change")
plt.xlabel("Inflation")
plt.ylabel("Exchange Rate Change")
plt.show()
```

Write one sentence interpreting the relationship.
