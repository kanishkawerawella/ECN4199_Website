## 3. Select Segmentation Variables


```python
features = ["Income", "Consumption", "Savings", "Household_Size", "Education_Years", "Digital_Access"]
X = data[features]
X.head()
```

## 4. Scale the Data


```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## 5. K-Means Clustering


```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
data["Cluster"] = kmeans.fit_predict(X_scaled)
data.groupby("Cluster")[features].mean().round(2)
```

## 6. Visualize Clusters


```python
plt.scatter(data["Income"], data["Savings"], c=data["Cluster"])
plt.xlabel("Income")
plt.ylabel("Savings")
plt.title("Household Segments")
plt.show()
```
