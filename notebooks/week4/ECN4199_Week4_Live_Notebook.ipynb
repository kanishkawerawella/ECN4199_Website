## 6. Create Analytical Variables

```python
data["Exchange_Rate_Change"] = data["Exchange_Rate"].pct_change() * 100
data["Recession"] = data["GDP_Growth"] < 0
data["Low_Reserves"] = data["Reserves"] < 4000

data
```

### Interpretation

Why is the first value of `Exchange_Rate_Change` missing? Should we automatically replace it with zero?

## 7. Classify Inflation

```python
def classify_inflation(rate):
    if rate < 5:
        return "Low"
    elif rate <= 15:
        return "Moderate"
    else:
        return "High"

data["Inflation_Category"] = data["Inflation"].apply(classify_inflation)
data[["Year", "Inflation", "Inflation_Category"]]
```

## 8. Test Boundary Values

```python
test_values = [4.9, 5, 15, 15.1]

for value in test_values:
    print(value, "->", classify_inflation(value))
```

### Boundary Interpretation

Explain why testing 5 and 15 matters for this classification rule.

## 9. Filter Important Economic Conditions

```python
high_inflation_years = data[data["Inflation_Category"] == "High"]
high_inflation_years
```

```python
recession_years = data[data["Recession"]]
recession_years
```

```python
multiple_stress_years = data[
    (data["Inflation_Category"] == "High") &
    (data["Recession"])
]

multiple_stress_years
```

### Interpretation

Which years show multiple macroeconomic stress signals?

## 10. Prepared Dataset for Week 5
