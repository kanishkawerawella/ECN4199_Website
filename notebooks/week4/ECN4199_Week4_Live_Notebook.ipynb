## 1. Load the Dataset

This notebook uses a small Sri Lanka macroeconomic dataset for practicing pandas operations.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/kanishkawerawella/ECN4199_Website/main/data/sri_lanka_macro.csv"
data = pd.read_csv(url)

data
```

```python
data.head()
data.info()
data.describe()
```

## 2. Select Variables

Select the variables needed for macroeconomic interpretation.

```python
data[["Year", "Inflation", "GDP_Growth", "Exchange_Rate"]]
```

## 3. Filter Economic Conditions

```python
high_inflation = data[data["Inflation"] > 10]
recession_years = data[data["GDP_Growth"] < 0]

high_inflation, recession_years
```

## 4. Create New Variables

```python
data["Exchange_Rate_Change"] = data["Exchange_Rate"].pct_change() * 100
data["High_Inflation"] = data["Inflation"] > 10

data
```

## 5. Sort and Interpret

Use sorting to identify the most important observations.

```python
data.sort_values(by="Inflation", ascending=False)
```

```python
data.sort_values(by="GDP_Growth")
```
