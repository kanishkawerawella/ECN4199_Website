## 3. Inspect Text Data


```python
data.head()
data["Topic"].value_counts()
```

## 4. Clean Text


```python
data["clean_text"] = data["Text"].str.lower().str.replace(r"[^a-z ]", "", regex=True)
data[["Text", "clean_text"]].head()
```

## 5. Count Common Terms


```python
vectorizer = CountVectorizer(stop_words="english")
term_matrix = vectorizer.fit_transform(data["clean_text"])
term_counts = pd.Series(term_matrix.toarray().sum(axis=0), index=vectorizer.get_feature_names_out()).sort_values(ascending=False)
term_counts.head(15)
```

```python
term_counts.head(10).sort_values().plot(kind="barh")
plt.title("Common Terms in Policy Texts")
plt.show()
```
