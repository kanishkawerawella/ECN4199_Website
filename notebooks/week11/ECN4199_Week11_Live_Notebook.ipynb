## 3. Prepare Time Series Data


```python
data["Date"] = pd.to_datetime(data["Date"])
data = data.sort_values("Date")
data["Time_Index"] = range(len(data))
data.head()
```

## 4. Plot the Series


```python
plt.plot(data["Date"], data["Inflation"], marker="o")
plt.title("Inflation Over Time")
plt.xlabel("Date")
plt.ylabel("Inflation")
plt.xticks(rotation=45)
plt.show()
```

## 5. Moving Average


```python
data["Inflation_MA3"] = data["Inflation"].rolling(3).mean()
plt.plot(data["Date"], data["Inflation"], label="Actual")
plt.plot(data["Date"], data["Inflation_MA3"], label="3-month moving average")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

## 6. Simple Trend Forecast


```python
train = data.iloc[:-6]
test = data.iloc[-6:]
model = LinearRegression()
model.fit(train[["Time_Index"]], train["Inflation"])
test = test.copy()
test["Forecast"] = model.predict(test[["Time_Index"]])
print("MAE:", round(mean_absolute_error(test["Inflation"], test["Forecast"]), 3))
test[["Date", "Inflation", "Forecast"]]
```

## 7. Forecast Plot


```python
plt.plot(train["Date"], train["Inflation"], label="Training")
plt.plot(test["Date"], test["Inflation"], label="Actual test")
plt.plot(test["Date"], test["Forecast"], label="Forecast")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```
