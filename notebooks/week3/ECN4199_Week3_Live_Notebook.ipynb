## 4. Conditional logic

```python
inflation = 12

if inflation < 5:
    classification = "Low inflation"
elif inflation <= 15:
    classification = "Moderate inflation"
else:
    classification = "High inflation"

print(classification)
```

### Boundary testing

```python
test_rates = [4.9, 5, 15, 15.1]

for rate in test_rates:
    if rate < 5:
        label = "Low inflation"
    elif rate <= 15:
        label = "Moderate inflation"
    else:
        label = "High inflation"

    print(rate, label)
```

## 6. Loops and stored results

```python
gdp_growth_rates = [3.5, 5.2, -7.8, 2.1, -1.4]
growth_categories = []

for growth in gdp_growth_rates:
    if growth < 0:
        category = "Recession"
    elif growth < 3:
        category = "Slow growth"
    else:
        category = "Strong growth"

    growth_categories.append(category)
    print(growth, category)

print(growth_categories)
```

## 8. Reusable functions

```python
def percentage_change(old_value, new_value):
    change = ((new_value - old_value) / old_value) * 100
    return change
```

```python
gdp_change = percentage_change(5000, 5400)
inflation = percentage_change(200, 240)
exchange_rate_change = percentage_change(300, 330)

print("GDP change:", round(gdp_change, 2))
print("CPI change:", round(inflation, 2))
print("Exchange-rate change:", round(exchange_rate_change, 2))
```

## 9. Classification function

```python
def classify_growth(growth):
    if growth < 0:
        return "Recession"
    elif growth < 3:
        return "Slow growth"
    else:
        return "Strong growth"
```

```python
for growth in gdp_growth_rates:
    print(growth, classify_growth(growth))
```

## 10. Integrated automation

```python
indicators = ["GDP", "CPI", "Exchange rate"]
old_values = [5000, 200, 300]
new_values = [5400, 240, 330]
interpretation_labels = ["Economic growth", "Inflation", "Currency depreciation"]

for indicator, old, new, label in zip(indicators, old_values, new_values, interpretation_labels):
    change = percentage_change(old, new)
    print(indicator, round(change, 2), label)
```
