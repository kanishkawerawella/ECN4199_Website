## Introduction to Machine Learning for Economics

This notebook introduces a simple supervised machine learning workflow. We use household characteristics to predict income, then evaluate how well the model performs on data it did not see during training.

## 1. Setup

We use `pandas` for data handling and `scikit-learn` for machine learning.

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error
```

## 2. Load the Dataset

The dataset is hypothetical and designed for classroom learning. Each row represents one household.

```python
url = "https://raw.githubusercontent.com/kanishkawerawella/ECN4199_Website/main/data/household_income.csv"
data = pd.read_csv(url)

data.head()
```

```python
data.describe()
```

## 3. Define Features and Target

In supervised learning, the features are the input variables. The target is the outcome we want to predict.

```python
features = ["Education", "Age", "Experience", "Urban"]
target = "Income"

X = data[features]
y = data[target]

X.head()
```

### Reflection

1. Which variables are features?
2. Which variable is the target?
3. Why might education and experience help predict income?

## 4. Split Training and Testing Data

The model learns from the training data. We evaluate it using the test data. This helps us check whether the model can generalize to unseen observations.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

print("Training observations:", len(X_train))
print("Testing observations:", len(X_test))
```

## 5. Model 1: Linear Regression

Linear regression is a useful starting point because it is easy to interpret.

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)

linear_predictions = linear_model.predict(X_test)
linear_mae = mean_absolute_error(y_test, linear_predictions)

print("Linear regression MAE:", round(linear_mae, 2))
```

```python
coefficients = pd.DataFrame({
    "Feature": features,
    "Coefficient": linear_model.coef_
})

coefficients
```

## 6. Model 2: Decision Tree

A decision tree can capture more flexible patterns, but it may overfit if it becomes too complex. We use `max_depth=3` to keep the tree simple.

```python
tree_model = DecisionTreeRegressor(max_depth=3, random_state=42)
tree_model.fit(X_train, y_train)

tree_predictions = tree_model.predict(X_test)
tree_mae = mean_absolute_error(y_test, tree_predictions)

print("Decision tree MAE:", round(tree_mae, 2))
```

## 7. Compare Predictions

Mean absolute error tells us the average size of prediction mistakes. Lower values mean better prediction performance for this metric.

```python
comparison = pd.DataFrame({
    "Actual Income": y_test.values,
    "Linear Prediction": linear_predictions,
    "Tree Prediction": tree_predictions
})

comparison["Linear Error"] = comparison["Actual Income"] - comparison["Linear Prediction"]
comparison["Tree Error"] = comparison["Actual Income"] - comparison["Tree Prediction"]

comparison.round(2)
```

### Reflection

1. Which model has the lower MAE?
2. Does lower prediction error automatically mean the model is better for policy decisions?
3. Which model is easier to explain to a non-technical audience?

## 8. Predict a New Household

Now create a new household and ask both models to predict income.

```python
new_household = pd.DataFrame({
    "Education": [14],
    "Age": [35],
    "Experience": [10],
    "Urban": [1]
})

print("Linear model prediction:", round(linear_model.predict(new_household)[0], 2))
print("Decision tree prediction:", round(tree_model.predict(new_household)[0], 2))
```
